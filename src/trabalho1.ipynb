```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# 1 
### a)

```python
np.random.seed(3483)
mi = 15
sigma2 = 6.7
n1 = 50
amostra_50 = np.random.normal(mi, np.sqrt(sigma2), size=(n1))
```

```python
plt.hist(amostra_50)
plt.title(f'Amostra de tamanho {n1}, mi = {mi}, sigma2 = {sigma2}')
plt.show()
```

```python
mi_chapeu = np.mean(amostra_50)
mi_chapeu
```

```python
sigma2_chapeu = np.sum((amostra_50 - mi_chapeu)**2)/n1
sigma2_chapeu
```

### b)

```python
np.random.seed(3483)
amostras_50 = np.random.normal(mi, np.sqrt(sigma2), size=(10, n1))
amostras_50.shape
```

```python
fig, ax = plt.subplots(2,5, figsize= (12,10))
ax_flat = ax.flatten()
for i in range(10):
    ax_flat[i].hist(amostras_50[i])
    ax_flat[i].set_title(f"Amostra {i + 1}")
fig.tight_layout()
plt.savefig('Amostras_50.jpeg')
```

```python
mi_chapeu_50 = np.mean(amostras_50, axis=1)
```

Cálculo de $\hat{\mu}$ para cada amostra de tamanho 50

```python
sigma2_chapeu_50 = np.sum((amostras_50 - mi_chapeu_50[:,None])**2, axis=1)/n1
```

Cálculo de $\hat{\sigma²}$ para cada amostra

```python
stats_50 = {
    "Amostra" : range(1,11),
    "Mi_chapeu" : mi_chapeu_50,
    "Sigma2_chapeu" : sigma2_chapeu_50,
}
```

```python
tabela_50 = pd.DataFrame(stats_50)
```

```python
tabela_50
```

```python
media_mi_50 = np.mean(mi_chapeu_50)
media_sigma2_50 = np.mean(sigma2_chapeu_50)
```

```python
print(media_mi_50)
print(media_sigma2_50)
```

### c)
Calculando as estatísticas empíricas

```python
vies_mi_50 = media_mi_50 - mi
vies_sigma2_50 = media_sigma2_50 - sigma2
```

```python
var_mi_50 = np.sum((mi_chapeu_50 - media_mi_50)**2)/10
```

```python
var_sigma2_50 = np.sum((sigma2_chapeu_50 - media_sigma2_50)**2)/10
```

```python
eqm_mi_50 = np.sum((mi_chapeu_50 - mi)**2)/10
```

```python
eqm_sigma2_50 = np.sum((sigma2_chapeu_50- sigma2)**2)/10
```

```python
print(f'eqm mi pela soma: {eqm_mi_50:.4f}')
print(f'eqm mi pela variância + vies**2: {var_mi_50 + vies_mi_50**2:.4f}')
```

### d)

```python
print(f'Média de mi_chapeu: {media_mi_50}, mi verdadeiro: {mi}')
print(f'Podemos ver que a diferença é o viés empírico: {vies_mi_50}')
```

```python
print(f'Média de sigma2_chapeu: {media_sigma2_50}, chapeu verdadeiro: {sigma2}')
print(f'Podemos ver que a diferença é o viés empírico: {vies_sigma2_50}')
```
